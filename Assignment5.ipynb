```python
# 3. Create a Pandas DataFrame using the dataset below: 
 
# Name Age Marks 
# Amit 20 85 
# Neha 21 90 
# Rahul 19 78 
# Sneha 22 88 
 
data = {
    "Name":["Amit","Neha","Rahul","Sneha"],             #pd.DataFrame()       
    "Age":[20,21,19,22],
    "Marks":[85,90,78,88]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n",df)
# a) Tasks: 
# ● Display first 2 rows 
print(df.head(2))                       # ● head()  
# ● Display only Name and Marks columns
print(df[["Name","Marks"]])              # ● Column Selection  
# ● Add a new column named “Result”
df["Result"] = ["Pass" if mark >= 40 else "Fail" for mark in df["Marks"]]
print(df)
# ● Display students with marks greater than 80  
print(df[df["Marks"] > 80])               # ● Conditional Filtering 
```

```python
# 4. Create a DataFrame using the dataset below: 
# Name       Age        Salary
# A          25         30000 
# B          NaN        40000 
# C          27         NaN 
# D          24         35000 

data = {
    "Name":["A","B","C","D"],
    "Age":[25,np.nan,27,24],
    "Salary":[30000,40000,np.nan,35000]
}
df_missing = pd.DataFrame(data)
print("Original Dataset:",df_missing)
# a) Tasks:                    # b) Expected Concepts: 
# ● Identify missing values  
print(df_missing.isnull())     # ● isnull() 
# ● Fill missing Age with average age 
avg_age = df_missing["Age"].mean()  # ● mean()
print(avg_age )
# ● Fill missing Salary with median salary  
median_salary = df_missing["Salary"].median()    # ● median() 
print(median_salary)
# ● Display cleaned dataset 
df_missing["Age"].fillna(avg_age, inplace=True)
df_missing["Salary"].fillna(median_salary, inplace=True)
print("Cleaned Dataset:\n",df_missing )           # ● fillna() 
```

```python
# 5. Dataset Link: Iris Dataset 
# a) Tasks:                                  # b) Expected Concepts: 
# ● Load CSV file using Pandas  
iris = pd.read_csv("Iris.csv")               # ● pd.read_csv() 
print(iris.head())
# ● Display dataset information  
print(iris.info())                           # ● info() 
# ● Find:  
# o Total rows and columns  
print("Rows :", iris.shape[0])
print("Columns :", iris.shape[1])
# o Average sepal length  
print("Average Sepal Length =", iris["SepalLengthCm"].mean())
# o Different species names  
print(iris["Species"].unique())              # ● unique() 
# ● Display summary statistics  
print(iris.describe())                       # ● describe()  
```
