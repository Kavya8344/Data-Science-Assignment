```python
# 1. Create a string variable with your name and perform the following operations: 
str="kavya"
# o Print the first character
print("First Character:",str[0])
# o Print the last character
print("Last Character:",str[4])
# o Print the length of the string 
print("Length of string:",len(str))
# o Convert the string to uppercase 
print("Upper Case:",str.upper())
# o Convert the string to lowercase 
print("Lower Case:",str.lower())
# o Reverse the string using slicing 
print("Reversed string:",str[::-1])
```

```python
# 2. Write a Python program to perform string slicing on a given string:
word= "stringslicing"
# o Print the first four characters 
print("First four characters:",word[:4])
# o Print characters from index 2 to 5 
print("Characters from 2 to 5:",word[2:5])
# o Print the complete string in reverse order
print("Reverse order:",word[::-1]) 
```

```python
# 3. Create a list containing numbers and perform the following operations: 
list=[2,4,4,5,7,6,6,8,12,12,12,14]
# o Add an element using append()
list.append(26)
print("Append element:",list)
# o Insert an element at a specific position
list.insert(3,76)
print("New list:",list)
# o Remove an element using remove()
list.remove(5) 
print("Remove element:",list)
# o Delete the last element using pop()
list.pop() 
print("Delete the last element:",list)
# o Reverse the list 
list.reverse()
print("Reverse list:",list)
# o Sort the list 
list.sort()
print("Sort list:",list)
# o Find the length of the list
print("Length of the list:",len(list)) 
# o Count the occurrence of an element 
print("Occurence of 12:",list.count(12))
```

```python
# 4. Create a tuple containing 5 subjects. Perform the following operations: 
tup=(45,89,78,32,59)
# o Print the first element 
print("First element:" ,tup[0])
# o Print the last element 
print("Last element:",tup[-1])
# o Print the length of the tuple 
print("Length of Tuple:",len(tup))
# o Print elements using slicing
print("Elements using slicing:",tup[0:5]) 
# o Find the maximum, minimum, and sum of numerical tuple elements
print("Maximum:",max(tup))
print("Minimum:",min(tup))
print("Sum:",sum(tup))
```

```python
#5. Write a Python program to demonstrate tuple packing and unpacking. Create a tuple of four values and assign them to four separate variables. 
# Tuple Packing
student = ("Kavya",10,"CS(AI)",95)
print("Packed Tuple:", student)
# Tuple Unpacking
name, roll_no, branch, marks = student
print("Name:", name)
print("Roll number:", roll_no)
print("Branch:", branch)
print("Marks:", marks)
```
