```python
#1. Find the largest, smallest, and second largest number from 10 user inputs.

nums=[]
for i in range(10):
    n=int(input("enter a integer: "))
    nums.append(n)
print(nums)
print("largest number: ",max(nums))
print("smallest number: ",min(nums))
nums.remove(max(nums))
print("second largest: ",max(nums))
```

```python
#2. Count vowels, consonants, digits, and special characters in a string.  
vowels=0
consonants=0
digits=0 
special=0
s=input("enter a string: ")
for i in s:
    if i.isalpha():
        if i in "aeiou":
            vowels+=1
        else:
            consonants+=1
    elif i.isdigit():
        digits+=1
    else:
        special+=1
print(s)
print("Vowels =", vowels)
print("Consonants =", consonants)
print("Digits =", digits)
print("Special Characters =", special)
```

```python
#3. Check whether a number is an Armstrong number. 
num = int(input("enter a number: "))
t = num
a = len(str(num))
sum = 0
while t > 0:
    digit = t % 10
    sum += digit ** a
    t //= 10
if sum == num:
    print(num, "is an armstrong number")
else:
    print(num, "is not an armstrong number")
```

```python
#4. Find duplicate elements and their frequency in a tuple.  
t = (12,45,75,54,45,12,65,84)
for i in set(t):
    if t.count(i) > 1:
       print("number:",i," frequency:",t.count(i))
```

```python
#5. Find the topper from a tuple of student names and marks. 
students = (("a", 65), ("b", 78), ("c", 84), ("d", 54), ("e", 99))
topper = students[0]
for s in students:
    if s[1] > topper[1]:
        topper = s
print("topper:", topper[0])
print("marks:", topper[1])
```

```python
#6. Count the frequency of each word in a sentence using a dictionary.  
sentence = input("Enter a sentence: ")
words = sentence.split()
freq = {}
for word in words:
    if word in freq:
        freq[word] = freq[word] + 1
    else:
        freq[word] = 1
print(freq)
```

```python
#7. Display students scoring above the average marks from a dictionary. 
students = {
    "a": 80,
    "b": 90,
    "c": 72,
    "d": 85
    }
total = 0
for marks in students.values():
    total = total + marks
avg = total / len(students)
print("average marks =", avg)
print("students above average:")
for name in students:
    if students[name] > avg:
        print(name, students[name])
```

```python
#9. Check whether a triangle is Equilateral, Isosceles, or Scalene. 
a = int(input("enter side 1: "))
b = int(input("enter side 2: "))
c = int(input("enter side 3: "))
if a == b == c:
    print("equilateral triangle")
elif a == b or b == c or a == c:
    print("isosceles triangle")
else:
    print("scalene triangle")
```
